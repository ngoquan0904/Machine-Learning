```python
import numpy as np
np.random.seed(42)
```

```python
means = [[2,2], [4,2]]
cov = [[.3, .2], [.2, .3]]
N = 10
X0 = np.random.multivariate_normal(means[0], cov, N).T # (2, N)
X1 = np.random.multivariate_normal(means[1], cov, N).T # (2, N)
X = np.concatenate((X0, X1), axis = 1) # (2, 2N)
y = np.concatenate((np.ones((1, N)), -1 * np.ones((1, N))), axis = 1)
#Xbar
X = np.concatenate((np.ones((1, 2*N)), X), axis=0) # (3, 2N)
```

```python
X, y
```

```python
def h(w, X):
  return np.sign(np.dot(w.T, X)) # (1, d) x (d, 2N) => (1,2N)

def has_coverged(X, y, w):
  return np.array_equal(h(w, X), y)

def perceptron(X, y, w_init):
  w = [w_init]
  N = X.shape[1]
  d = X.shape[0]
  mis_points = []
  while True:
    # shuffle data, ví dụ array([1, 7, 4, 2, 8, 6, 5, 0, 9, 3])
    mix_id = np.random.permutation(N)
    for i in range(N):
      xi = X[:, mix_id[i]].reshape(d, 1)
      yi = y[0, mix_id[i]]
      # (1, d) x (d, 1) => (1,1)
      if h(w[-1], xi)[0] != yi:
        mis_points.append(mix_id[i])
        w_new = w[-1] + yi * xi
        w.append(w_new)
    # lặp cho đến khi không còn điểm nào bị misclassified
    if has_coverged(X, y, w[-1]):
      break
  return (w, mis_points)
```

```python
d = X.shape[0]
w_init = np.random.randn(d, 1)
(w, m) = perceptron(X, y, w_init)
```
