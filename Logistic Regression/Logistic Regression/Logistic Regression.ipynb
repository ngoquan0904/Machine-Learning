```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
from sklearn.metrics import accuracy_score
from tensorflow.keras.datasets import mnist
```

```python
(X_train, y_train), (X_test, y_text) = mnist.load_data()
```

```python
X_train.shape, y_train.shape
```

```python
# Flattening
X_train_all = X_train.reshape(X_train.shape[0], -1)
X_test_all = X_test.reshape(X_test.shape[0], -1)
```

```python
X_train_all[0].shape
```

```python
cls = [[0], [1]]
def extract_data(X, y, classes):
  '''
    Lấy ra các điểm dữ liệu có nhãn trong classes
  '''
  y_res_id = np.array([]) # mặc định kdl float64
  # Lấy ra index của ảnh có nhãn là 0
  for i in cls[0]:
      y_res_id = np.hstack((y_res_id, np.where(y == i)[0]))
  n0 = len(y_res_id)
  # Lấy ra index của ảnh có nhãn là 0
  for i in cls[1]:
      y_res_id = np.hstack((y_res_id, np.where(y == i)[0]))
  n1 = len(y_res_id) - n0
  y_res_id = y_res_id.astype(int)
  X_res = X[y_res_id, :]/255.0
  y_res = np.asarray([0]*n0 + [1]*n1)
  return (X_res, y_res)
```

```python
X_train, y_train = extract_data(X_train_all, y_train, cls)
X_test, y_test = extract_data(X_test_all, y_text, cls)
```

```python
logreg = linear_model.LogisticRegression(C=1e5)
logreg.fit(X_train, y_train)
y_pred = logreg.predict(X_test)
accuracy_score(y_test, y_pred)
```
